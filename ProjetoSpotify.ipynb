# **Criando a variável target**🎯

De acordo com a regra de negócio, a coluna valence é uma medida de **0,0 a 1,0 que descreve a positividade musical transmitida por uma faixa**. Faixas com **alta valência soam mais positivas** (por exemplo, feliz, alegre, eufórica), enquanto faixas com **baixa valência** soam mais negativas (por exemplo, triste, deprimida, irritada).
Para criar a coluna alvo do modelo preditivo, basta utizar a coluna "valence" para ser o critério de músicas agitadas ou lentas.

```python
import matplotlib.pyplot as plt
#Analisando a coluna "valence" para criar a nossa target
plt.hist(df['valence'], bins=20, color='blue', edgecolor='black')
plt.xlabel('Valence')
plt.ylabel('Frequência')
plt.title('Histograma da Coluna Valence')
plt.show()
```

```python
df['valence'].describe()
```

```python
def categorizar_valence(row):
  if row['valence'] > 0.5:
    return 'agitada'
  else:
    return 'lenta'

#Crie a nova coluna 'target'usando a função categorizar_valence
df['target'] = df.apply(categorizar_valence, axis=1) #O apply vai aplicar essa função para cada uma das linhas que temos no dataset e para criar a nova coluna também
df.head()
```

# Tratando os dados categóricos 🅰️

LabelEncoder: Essa classe é utilizada para codificar rótulos de classes em números inteiros. É frequentemente usado quando se trabalha com algoritmos de aprendizado supervisionado que requerem rótulos numéricos.

```python
def label_encoder_dataframe(df, columns_to_encode):
  from sklearn.preprocessing import LabelEncoder
  le = LabelEncoder()

  for column in columns_to_encode:
    if column in df.columns:
      df[column] = le.fit_transform(df[column])
    else:
      print('A lista possui colunas que não existem no DataFrame.')

  return df

colunas_a_codificar = ['artists', 'album_name', 'track_name', 'explicit', 'track_genre', 'target']
label_encoder_dataframe(df_musica, colunas_a_codificar)
df_musica.head()
```

# Separando os dados em treino e teste 🔄

```python
df_musica.columns
```

```python
#Separando os dados em treino e teste (método hold out)
from sklearn.model_selection import train_test_split #Ferramenta base que trabalha separando o treino e teste, treino para reconhecer os padrões 80/70% da base, teste para testar em dados
                                                     #não vistos, mais ou menos 20/30%.

#X = df_musica.drop('target', axis=1)
X = df_musica[['popularity', 'duration_ms', 'danceability', 'energy', 'loudness', 'acousticness', 'instrumentalness', 'liveness', 'track_genre']] #Aqui são as variáveis características
                                                                                                                                                 #, ou independentes.
y = df_musica['target'] #A resposta é o y, que no caso é a target.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=7)
#Separando as variáveis dessa forma da para armazenar cada conjuntinho de teste
#Depois dentro do train_test_split primeiro você chama as variáveis características e depois e depois a dependente.
#Test size 0.20 porque siginifica que é 80% para treinar e 20% para testar
#stratify é legal principalmente quando você tem uma base que não está tão desiquilibrada para poder distribuir de forma estratificada, para não ter meio que desequilibrio na hora
#de colocar em treino e teste.
#random_state é uma semente aleatória que a gente vai por no dado e manter ela sempre em toda execução dos nossos algoritmos, isso significa que sempre que você reproduzir esse
#algoritmo vai estar nesse mesmo cenário, para não deixar enviesado, porque isso muda também resultados.
```

```python
print(X_train.shape, X_test.shape)
```

# Normalizando os dados 📏
Ao aplicar o MinMaxScaler, todos os valores dos dados serão transformados para o intervalo entre 0 e 1, onde o valor mínimo será 0 e o valor máximo será 1. Essa técnica é especialmente útil quando os algoritmos de aprendizado de máquina são sensíveis à escala dos dados.

```python
from sklearn.preprocessing import MinMaxScaler

#Criar uma instância do MinMaxScaler
scaler = MinMaxScaler()

scaler.fit(X_train) #Você faz o fit só na base de treinamento porque, se você fazer esse fit significa você treinar ali as características dessa base de treino, se você fazer o treino
                    #dela e em cima da base de teste também, você está fazendo com que as estatísticas dessa base de treino e teste não sirvam pra nada, ela vai estar se replicando
                    #então a ideia aqui é deixa eu entender o comportamente da minha base de treinamento e eu preciso generalizar novos dados na base de teste, então para não ter
                    #vazamento dessas estatísticas, desse comportamento, que é o Data Leak, a gente fez o fit só na base de treino

X_train_escalonado = scaler.transform(X_train) #A transformação é aplicado as duas bases, a de treino, e a de teste que são os dados do mundo real.
X_test_escalonado = scaler.transform(X_test) #E a criação dessas duas novas variáveis se dá para armazenar uma nova representação, para não criar confusão no teste, isso vai ser a nova
                                              #versão entre 0 e 1 de cada uma das bases.
```

# Criando uma função para executar modelos de machine learning 🚀

Criação de uma função que vai executar o modelo preditivo e avaliar a sua performance utilizando as principais métricas de validação

```python
def roda_modelo(modelo):

  from sklearn.metrics import roc_curve, roc_auc_score, classification_report

  #Treinando modelo com os dados de treino
  modelo.fit(X_train_escalonado, y_train)

  #Calculando a probabilidade e calculando o AUC
  prob_predic = modelo.predict_proba(X_test_escalonado) #Obter as probabilidades associadas às classes previstas para cada inst6ancia de dados
  auc = roc_auc_score(y_test, prob_predic[:, 1])
  print(f"AUC {auc}")

  #Fazendo a predição dos dados de teste e calculando o classification report
  predicao = modelo.predict(X_test_escalonado)
  print("\nClassification Report")
  print(classification_report(y_test, predicao))

  print("\nRoc Curve\n")
  #Fazer previsões de probabilidade
  y_pred_probs = modelo.predict_proba(X_test_escalonado)[:, 1]

  #Calcular a curva ROC
  fpr, tpr, thresholds = roc_curve(y_test, y_pred_probs)

  #Calcular a AUC (área sobre a curva ROC)
  auc = roc_auc_score(y_test, y_pred_probs)

  #Plotar a curva ROC
  plt.figure(figsize=(8, 6))
  plt.plot(fpr, tpr, color='blue', label=f'ROC curve (AUC = {auc:.2f})') #linewidth
  plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
  plt.xlabel('Taxa De Falso Positivo')
  plt.ylabel('Taxa De Verdadeiro Positivo')
  plt.title('Curva ROC')
  plt.legend(loc='lower right')
  plt.show()

  #Converter probabilidades em classes preditas (0 ou 1)
  y_pred = (y_pred_probs > 0.5).astype(int)
```

# Regressão logística
O modelo se baseia em uma **função logística**, que transforma as variáveis independentes em uma probabilidade entre 0 e 1.  Para novas entradas de dados, o modelo calcula a **probabilidade do evento binário ocorrer**.

```python
from sklearn.linear_model import LogisticRegression
modelo_logistico = LogisticRegression()
roda_modelo(modelo_logistico)
#Quanto mais a curva estiver próxima do 1, melhor é, pois, significa que o meu classficador consegue distinguir bem as classes, quanto mais longe, pior é, nesse caso o modelo não ficou
#banaca, as outras taxas estão baixas também
```

# KNN (K-Nearest Neighbors)
Para um novo ponto de dados, o KNN identifica os **K pontos mais próximos (vizinhos)** no conjunto de treinamento. A classe do novo ponto é a classe mais frequente entre os K vizinhos.

```python
from sklearn.neighbors import KNeighborsClassifier
modelo_knn = KNeighborsClassifier(n_neighbors=3) #Uma estratégia legal é colocar números impáres de K para não ter empate
roda_modelo(modelo_knn)
#Não ficou legal também
```

# Random Florest
O Random Forest Classifier é um algoritmo de **ensemble learning**, que **combina vários modelos para melhorar a performance**. O modelo cria uma floresta de árvores de decisão, onde cada árvore é treinada em um subconjunto aleatório dos dados (bootstrap). A classe final do novo ponto de dados é a classe mais votada pelas árvores da floresta.

```python
from sklearn.ensemble import RandomForestClassifier
modelo_random_forest = RandomForestClassifier(max_depth=7, n_estimators=100)
#max_depth é a profundidade da árvore, seria o quanto de ramificações vai criar na árvore
#n_estimators seria o número de árvores que vão estar na minha floresta
roda_modelo(modelo_random_forest)
#ainda não ficou legal o modelo nesse caso!
```

# Testando novos parâmetros com Grid Search

O GridSearch é uma técnica para criar um dicionário de parâmetros e você pode juntar, fazer um teste de vários conjuntinhos, que a gente fala que são a validação cruzada, ou seja, vários cenários, várias partes aleatórias da base de dados e faz a tentativa com vários tipos de parâmetros

```python
from sklearn.model_selection import GridSearchCV

# Defina os parâmetros a serem testados
param_grid = {
    "n_estimators": [100, 200, 300, 400, 500],
    "max_depth": [5, 10, 15, 20, 25]
}

#GridSearchCV
grid_search = GridSearchCV(RandomForestClassifier(), param_grid, cv=25, scoring='f1', n_jobs=1)
#Lembrando f1 score é a média harmônica entre precisão e revocação
#O cv significa que ele testa em cinco caixinhas aleatórias, fazendo um cenário diferentes ali dos dados, generalizando

#Ajuste o modelo ao conjunto de dados
grid_search.fit(X_train_escalonado, y_train)

rf_params = grid_search.best_params_
print("Melhores Hiperparâmetros:", rf_params)
```
